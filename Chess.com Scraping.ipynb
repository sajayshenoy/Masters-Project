```python
import requests
import time
```

```python
headers = {"User-Agent": "Mozilla/5.0"}
```

```python
url = "https://api.chess.com/pub/player/magnuscarlsen"
```

```python
response = requests.get(url, headers=headers)
print(response.text)
```

```python
url_gms = "https://api.chess.com/pub/titled/GM"
response = requests.get(url_gms, headers=headers)
gms = response.json()["players"]
```

```python
for name in gms[:5]:
    url = f"https://api.chess.com/pub/player/{name}"
    response = requests.get(url, headers=headers)
    print(response.text)
```
