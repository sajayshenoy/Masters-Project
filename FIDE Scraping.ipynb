```python
import requests
import fideparser
import pandas as pd
```

```python
def get_info(fide_id):
    url = f"{"https://api.chesstools.org"}/fide/{fide_id}"
    response = requests.get(url)
    return response.json()
```

```python
def get_history(fide_id):
    url = f"{"https://api.chesstools.org"}/fide/player_history/"
    params = {"fide_id": fide_id}
    response = requests.get(url, params=params)
    return response.json()
```

```python
ids = []
with open("standard_rating_list.txt", encoding="utf-8") as f:
    next(f)
    for line in f:
        parts = line.strip().split()
        if parts:
            ids.append(parts[0])
```

```python
ids
```

```python
for fid in ids[:5]:
    print("Player Info:", get_info(fid))
    print("Rating History:", get_history(fid))
```

```python
#testing data storage for 50 ids
data = []
for fid in ids[:50]:
    info = get_info(fid)
    history = get_history(fid)
    data.append({"fide_id": fid, "name": info.get("name"), "federation": info.get("fed"),
        "title": info.get("title"), "current_rating": info.get("std_rating"), "history": history})

df = pd.DataFrame(data)
df.to_excel("chess_players.xlsx", index=False)
```

```python
import ast

df["history"] = df["history"].apply(lambda x: ast.literal_eval(str(x)))

df["latest_date"] = df["history"].apply(lambda h: h[-1]["date"] if h else None)
df["classical_rating"] = df["history"].apply(lambda h: h[-1].get("classical_rating") if h else None)
df["rapid_rating"] = df["history"].apply(lambda h: h[-1].get("rapid_rating") if h else None)
df["blitz_rating"] = df["history"].apply(lambda h: h[-1].get("blitz_rating") if h else None)

df.to_excel("chess_players_split.xlsx", index=False)
```

```python
rows = []
for _, r in df.iterrows():
    history_list = ast.literal_eval(str(r["history"]))
    for entry in history_list:
        rows.append({"fide_id": r["fide_id"], "name": r["name"], "date": entry.get("date"), "classical_rating": entry.get("classical_rating"),
                     "rapid_rating": entry.get("rapid_rating"), "blitz_rating": entry.get("blitz_rating")})

history_df = pd.DataFrame(rows)
history_df.to_excel("chess_players_history_expanded.xlsx", index=False)
```
