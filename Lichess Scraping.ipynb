```python
from dotenv import load_dotenv
import os

load_dotenv()
api_key = os.getenv("lichess_api_key")
```

```python
import berserk
```

```python
session = berserk.TokenSession(api_key)
client = berserk.Client(session=session)
```

```python
client.account.get()
```

```python
client.users.get_all_top_10()
```

```python
lb = client.users.get_leaderboard('blitz', count = 5)
```

```python
for player in lb:
    username = player['username']
    ratings = client.users.get_rating_history(username = username)
    print(f"Rating history for {username}:\n{ratings}")
```
